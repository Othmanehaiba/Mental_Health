```python
import pandas as pd
import numpy as np

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv("../data/raw/student_mental_health.csv")

X = df.drop(columns=[
    "Mental_Health_Risk",
    "Student_ID"
])

y = df["Mental_Health_Risk"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

numerical_cols = X_train.select_dtypes(
    include=np.number
).columns

categorical_cols = X_train.select_dtypes(
    exclude=np.number
).columns

display(X_train, X_test, y_train, y_test)
```

```python
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numerical_pipeline, numerical_cols),
    ("cat", categorical_pipeline, categorical_cols)
])
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=2000, random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42)
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
results = []

for name, model in models.items():
    pipeline = Pipeline([
        ("preprocessing", preprocessor),
        ("model", model)
    ])
    
    cv_scores = cross_validate(
        pipeline, X_train, y_train, cv=cv,
        scoring=["accuracy", "precision_macro", "recall_macro", "f1_macro"]
    )
    
    results.append({
        "Model": name,
        "Accuracy": round(cv_scores["test_accuracy"].mean(), 4),
        "Precision (Macro)": round(cv_scores["test_precision_macro"].mean(), 4),
        "Recall (Macro)": round(cv_scores["test_recall_macro"].mean(), 4),
        "F1 Macro": round(cv_scores["test_f1_macro"].mean(), 4)
    })

df_comparison = pd.DataFrame(results).sort_values(by="F1 Macro", ascending=False)
display(df_comparison)
```

```python
from sklearn.model_selection import GridSearchCV

lr_pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LogisticRegression(max_iter=2000, random_state=42))
])

param_grid_lr = {
    "model__C": [0.01, 0.1, 1.0, 10.0],
}

grid_lr = GridSearchCV(lr_pipeline, param_grid_lr, cv=5, scoring="f1_macro", n_jobs=-1)
grid_lr.fit(X_train, y_train)

print("Best LR Params:", grid_lr.best_params_)
print("Best LR CV F1 Macro:", round(grid_lr.best_score_, 4))

rf_pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", RandomForestClassifier(random_state=42))
])

param_grid_rf = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [5, 10],
    "model__min_samples_split": [10, 20]
}

grid_rf = GridSearchCV(rf_pipeline, param_grid_rf, cv=5, scoring="f1_macro", n_jobs=-1)
grid_rf.fit(X_train, y_train)

print("Best RF Params:", grid_rf.best_params_)
print("Best RF CV F1 Macro:", round(grid_rf.best_score_, 4))

best_model = grid_lr.best_estimator_ if grid_lr.best_score_ >= grid_rf.best_score_ else grid_rf.best_estimator_
print("\nSelected Final Model:", type(best_model.named_steps["model"]).__name__)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = best_model.predict(X_test)

print("Test Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred, labels=best_model.classes_)
plt.figure(figsize=(6, 4))
sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=best_model.classes_,
    yticklabels=best_model.classes_
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix (Test Set)")
plt.show()
```

```python
import os
import joblib

error_df = X_test.copy()
error_df["Actual"] = y_test
error_df["Predicted"] = y_pred
misclassified = error_df[error_df["Actual"] != error_df["Predicted"]]

print(f"Total misclassified test samples: {len(misclassified)} / {len(X_test)}")
display(misclassified.head(10))

os.makedirs("../models", exist_ok=True)
joblib.dump(best_model, "../models/best_model.pkl")
print("Final pipeline saved to ../models/best_model.pkl")
```
